Importing the Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn import metrics
from xgboost import XGBRegressor
```

Importing the California House Price Dataset

```python
house_price_dataset = fetch_california_housing()
```

```python
print(house_price_dataset)
```

```python
#loading the dataset into pandas dataframe
house_price_data = pd.DataFrame(house_price_dataset.data, columns = house_price_dataset.feature_names)
```

```python
#printing first 5 rows of data
house_price_data.head()
```

```python
house_price_data['Price'] = house_price_dataset.target
```

```python
#printing first 5 rows of data after adding the column price
house_price_data.head()
```

```python
#no. of rows and cols in the dataset
house_price_data.shape
```

```python
#check for missing value
house_price_data.isnull().sum()
```

```python
#statistical measures of the dataset
house_price_data.describe()
```

Understanding the correaltion between the various features in the dataset

1. Positive Correaltion

2. Negative Correlation

```python
correaletion = house_price_data.corr()
```

```python
#constructing the heatmap to understand the correalation
plt.figure(figsize=(10,10))
sns.heatmap(correaletion, cbar=True, square=True, fmt='.1f', annot=True, annot_kws={'size':8}, cmap="Blues")
```

Splitting the data and target

```python
X = house_price_data.drop(['Price'], axis=1)
Y = house_price_data['Price']
```

```python
print(X)
```

```python
print(Y)
```

Splitting data into training data and test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training using XGBoost Regressor

```python
#loading the model
model = XGBRegressor()
```

```python
#training the model with X_train
model.fit(X_train, Y_train)
```

Evaluation of Model

Prediction on training data

```python
#accuracy on prediction of training data
training_data_prediction  = model.predict(X_train)
```

```python
print(training_data_prediction)
```

```python
#R squared error
score_1 = metrics.r2_score(Y_train, training_data_prediction)

#Mean Absolute error
score_2 = metrics.mean_absolute_error(Y_train, training_data_prediction)

print('R squared error: ',score_1)
print('Mean Absolute error: ',score_2)
#If the score is low it means model is working fine
```

Visualising the actual prices and predicted prices

```python
plt.scatter(Y_train, training_data_prediction)
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.show()
```

Prediciton on test data

```python
#accuracy on prediction of test data
test_data_prediction  = model.predict(X_test)
```

```python
print(test_data_prediction)
```

```python
#R squared error
score_1 = metrics.r2_score(Y_test, test_data_prediction)

#Mean Absolute error
score_2 = metrics.mean_absolute_error(Y_test, test_data_prediction)

print('R squared error: ',score_1)
print('Mean Absolute error: ',score_2)
```
